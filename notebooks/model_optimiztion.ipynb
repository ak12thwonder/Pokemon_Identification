```python
from tensorflow.keras.models import load_model

# Path to your saved model file
model = load_model(r'..\backend\models\pokemon_resnet_model.keras')
```

```python
test_dir=r"..\data\processed\pokemon_images_255_test"
```

```python
test_ds = tf.keras.utils.image_dataset_from_directory(
    test_dir,
    labels='inferred',         # or 'categorical' for one-hot
    label_mode='categorical',  # or 'int'
    batch_size=32,
    image_size=(128, 128),
    shuffle=False
)
```

```python
import numpy as np
y_true = []
y_pred = []

for images, labels in test_ds:
    preds = model.predict(images)
    y_true.extend(np.argmax(labels.numpy(), axis=1))
    y_pred.extend(np.argmax(preds, axis=1))
```

```python

import tensorflow as tf

train_dir = "../data/processed/pokemon_images_255_class"  # or your actual training directory
img_height, img_width = 128, 128

# Use batch_size=1 and shuffle=False for speed and consistency
ds = tf.keras.utils.image_dataset_from_directory(
    train_dir,
    labels='inferred',
    label_mode='categorical',
    batch_size=1,
    image_size=(img_height, img_width),
    shuffle=False
)
class_names = ds.class_names
print(class_names)
```

```python
import json
with open("class_names.json", "w") as f:
    json.dump(class_names, f)
```

```python
import json
with open("class_names.json", "r") as f:
    class_names = json.load(f)
```

```python
print(classification_report(y_true, y_pred, target_names=class_names))
```

```python
macro_precision = precision_score(y_true, y_pred, average='macro')
macro_recall = recall_score(y_true, y_pred, average='macro')
micro_precision = precision_score(y_true, y_pred, average='micro')
micro_recall = recall_score(y_true, y_pred, average='micro')

print(f"Macro Precision: {macro_precision:.4f}, Macro Recall: {macro_recall:.4f}")
print(f"Micro Precision: {micro_precision:.4f}, Micro Recall: {micro_recall:.4f}")
```

```python
from PIL import Image
import numpy as np

img_path = r"..\data\processed\pokemon_images_255_test\Pikachu\Pikachu.png"  # Change to your image path
img_height, img_width = 128, 128  # Use the same size as during training

# Load and preprocess the image
image = Image.open(img_path).convert("RGB").resize((img_height, img_width))
img_array = np.array(image)  # Do NOT divide by 255 if you used Rescaling layer
img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension
```
