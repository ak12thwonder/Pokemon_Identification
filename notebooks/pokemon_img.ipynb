```python
import pandas as pd
import numpy as np 
```

```python
import os

# Path to your flattened image directory
root_dir = "../data/processed/pokemon_images_train"

species_counts = {}

for species in os.listdir(root_dir):
    species_dir = os.path.join(root_dir, species)
    if os.path.isdir(species_dir):
        # Count image files in the species folder
        num_images = sum(
            fname.lower().endswith(('.png', '.jpg', '.jpeg'))
            for fname in os.listdir(species_dir)
        )
        species_counts[species] = num_images
uniq_img=set()
# Print the results
for species, count in sorted(species_counts.items(), key=lambda x: x[1], reverse=True):
    print(f"{species}: {count} images")
    uniq_img.add(count)

# Optionally, get summary statistics
total_species = len(species_counts)
total_images = sum(species_counts.values())
print(f"\nTotal species: {total_species}")
print(f"Total images: {total_images}")
```
